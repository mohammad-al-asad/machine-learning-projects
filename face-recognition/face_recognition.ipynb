```python
import torch
import torch.nn as nn
import torchvision.models as models
import numpy as np
import cv2
import face_recognition
from mtcnn import MTCNN
from google.colab import drive

drive.mount('/content/drive')
```

```python
def encode_face(image_path):
    image = face_recognition.load_image_file(image_path)
    encodings = face_recognition.face_encodings(image)
    if len(encodings) == 0:
        return None

    return encodings[0]
```

```python
my_image = '/content/drive/My Drive/AI-ML/Colab Notebooks/face_recognizer/my_image.png'

known_faces = []
known_names = []
known_faces.append(encode_face(my_image))
known_names.append("Asad")
```

```python
def recognize(image_path):
    emb = encode_face(image_path)

    if emb is None:
        return "No face found"

    distances = np.linalg.norm(known_faces - emb, axis=1)
    min_index = np.argmin(distances)

    if distances[min_index] < 0.6:
        return f"Welcome {known_names[min_index]}"
    else:
        return "Unknown"
```

```python
my_image2 = '/content/drive/My Drive/AI-ML/Colab Notebooks/face_recognizer/my_image2.jpg'
recognize(my_image2)
```
